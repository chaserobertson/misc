You are working for a fruit and vegetable vendor who has taken over her grandfather's business.
She would like to help some of your customers to identify the best way to pay their bill. The first
step is identifying how many different ways a customer can pay for a given bill without getting
any change back. We need you to come up with a solution for the vendor's system to identify
how many ways a bill can be paid using coin currency. Assume the only available payment
method is with coins of $1, $2, or $5.

Write a function that takes in a single integer (total bill) and returns an integer which indicates
the amount of different ways a customer could pay for the bill. The order matters in this problem,
so count each unique order as a unique way to pay (See example below).

Example:
- Bill: $3
- Possible ways to pay: 3
- ($1,$1,$1)
- ($1, $2)
- ($2,$1)

Requirements:
- Your preferred text editor
- Your solution must be in a function.
- Your function must be named "waysToPay"
- The input for waysToPay must an integer.
- The output for waysToPay must be an integer

Follow up questions:
- Can you write it where the order doesn't matter?
($1,$2) and ($2,$1) are counted as the same
- Can you write it with/without recursion?

```python
from functools import cache

# Calculates the number of ways to pay a bill using specified coins
# Always returns at least 1, because paying no coins is still a way to pay
def waysToPay(bill: int, ordered: bool = False, coins: set[int] = {1, 2, 5}) -> int:
    # recursive function to find all payment permutations
    # cached to enable reuse of intermediate results, necessary for larger bills
    @cache
    def payPermutations(bill: int) -> list[int]:
        # base case: no bill, no way to pay
        if bill < 1:
            return [tuple()]
        
        # use a set to enable ordered permutations
        ways_to_pay = set()

        # try to use each coin
        for coin in coins:
            # skip this coin if it'd result in overpayment
            if bill - coin < 0:
                continue
            # recurse on remaining bill
            for permutation in payPermutations(bill - coin):
                way_to_pay = (coin,) + permutation
                # sort if results should be ordered
                ways_to_pay.add(tuple(sorted(way_to_pay)) if ordered else way_to_pay)

        return ways_to_pay
    
    pay_ways = payPermutations(bill)
    print(pay_ways)
    return len(pay_ways)

waysToPay(3), waysToPay(3, ordered=True), waysToPay(5), waysToPay(5, ordered=True), waysToPay(0), waysToPay(-1), waysToPay(100, ordered=True)
```
